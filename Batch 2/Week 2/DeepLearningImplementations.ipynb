## 2. Step 1: Artificial Neural Network (ANN / Multi-Layer Perceptron)

- **Architecture:** Flatten -> Dense(128, ReLU) -> Dense(64, ReLU) -> Dense(10, Softmax)
- **Loss:** Sparse Categorical Crossentropy | **Optimizer:** Adam


```python
from Step_1_ANN import ArtificialNeuralNetwork

ann = ArtificialNeuralNetwork(epochs=10, batch_size=32)
ann_results = ann.run()
print("\nANN Results Summary:", ann_results)
```

## 3. Step 2: Convolutional Neural Network (CNN)
**Task:** Spatial feature extraction & digit classification on the MNIST dataset.
- **Architecture:** Conv2D(32, 3x3, ReLU) -> MaxPool(2x2) -> Conv2D(64, 3x3, ReLU) -> MaxPool(2x2) -> Flatten -> Dense(128, ReLU) -> Dropout(0.5) -> Dense(10, Softmax)
- **Loss:** Sparse Categorical Crossentropy | **Optimizer:** Adam


```python
from Step_2_CNN import ConvolutionalNeuralNetwork

cnn = ConvolutionalNeuralNetwork(epochs=10, batch_size=64)
cnn_results = cnn.run()
print("\nCNN Results Summary:", cnn_results)
```

## 4. Step 3: Recurrent Neural Network (SimpleRNN)
**Task:** Sequential sentiment analysis on the IMDB Movie Reviews dataset.
- **Preprocessing:** Vocabulary size = 10,000, Max Sequence Length = 200 tokens (padded)
- **Architecture:** Embedding(10000 -> 32) -> SimpleRNN(64) -> Dense(1, Sigmoid)
- **Loss:** Binary Crossentropy | **Optimizer:** Adam


```python
from Step_3_RNN import RecurrentNeuralNetwork

rnn = RecurrentNeuralNetwork(epochs=5, batch_size=64)
rnn_results = rnn.run()
print("\nSimpleRNN Results Summary:", rnn_results)
```

---
## 5. Step 4: Long Short-Term Memory (LSTM)
**Task:** Long-range temporal sentiment analysis on the IMDB Movie Reviews dataset.
- **Architecture:** Embedding(10000 -> 64) -> LSTM(64) -> Dense(1, Sigmoid)
- **Gating Mechanism:** Forget Gate ($f_t$), Input Gate ($i_t$), Candidate Cell ($\tilde{C}_t$), Output Gate ($o_t$)
- **Loss:** Binary Crossentropy | **Optimizer:** Adam


```python
from Step_4_LSTM import LSTMNeuralNetwork

lstm = LSTMNeuralNetwork(epochs=5, batch_size=64)
lstm_results = lstm.run()
print("\nLSTM Results Summary:", lstm_results)
```
